```python
import matplotlib.pyplot as plt

def draw_neural_network(input_labels, hidden_layers, output_labels):
    layer_sizes = [len(input_labels)] + hidden_layers + [len(output_labels)]
    layer_x_positions = range(len(layer_sizes))

    fig, ax = plt.subplots(figsize=(12, 6))
    ax.set_title("Neural Network Architecture for Credit Card Fraud Detection", fontsize=14)

    # Store neuron positions
    neuron_positions = []

    for layer_idx, layer_size in enumerate(layer_sizes):
        y_positions = [i - layer_size / 2 for i in range(layer_size)]
        x_positions = [layer_x_positions[layer_idx]] * layer_size
        neuron_positions.append(list(zip(x_positions, y_positions)))

        # Draw neurons
        ax.scatter(x_positions, y_positions, s=1000, zorder=3)

        # Label input and output neurons
        if layer_idx == 0:
            for i, label in enumerate(input_labels):
                ax.text(x_positions[i] - 0.1, y_positions[i], label,
                        ha='right', va='center', fontsize=9)
        elif layer_idx == len(layer_sizes) - 1:
            for i, label in enumerate(output_labels):
                ax.text(x_positions[i] + 0.1, y_positions[i], label,
                        ha='left', va='center', fontsize=9)

    # Draw connections
    for layer_idx in range(len(neuron_positions) - 1):
        for (x1, y1) in neuron_positions[layer_idx]:
            for (x2, y2) in neuron_positions[layer_idx + 1]:
                ax.plot([x1, x2], [y1, y2], color='gray', linewidth=0.7)

    ax.axis('off')
    plt.tight_layout()
    plt.show()


# Input features
input_features = [
    "Transaction Amount",
    "Transaction Time",
    "Merchant Category",
    "Customer Age",
    "Account Balance",
    "Transactions Today"
]

# Hidden layers
hidden_layers = [8, 4]

# Output
output_labels = ["Fraud Probability\n(0 = Genuine, 1 = Fraud)"]

# Draw the network
draw_neural_network(input_features, hidden_layers, output_labels)
```
